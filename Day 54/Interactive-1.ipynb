```python
## Python decorator function

def decorator_function(function):
    def wrapper_function():
        function()
    return wrapper_function

decorator_function(2+2)
```

```python
import time
    
def decorator_function(function):
    def wrapper_function():
        time.sleep(2)
        function()
    return wrapper_function
```

```python
import time

def say_hello():
    time.sleep(2)
    print("Hello")
```

```python
say_hello()
```

```python
@decorator_function
def say_hello():
    print("Hello")
```

```python
say_hello()
```

```python
@decorator_function
def say_bye():
    print("Bye")
```

```python
say_bye()
```

```python
def say_greeting():
    print("How are you?")
```

```python
decorated_function = decorator_function(say_greeting)
decorated_function()
```
