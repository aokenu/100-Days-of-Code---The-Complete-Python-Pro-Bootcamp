```python
def tea_order(customer_name, tea_type, *args, **kwargs):
    print(customer_name, "ordered a",tea_type, "tea")
    for arg in args:
        print(" -Add:", arg)
    
    for key, value in kwargs.items():
        print(" -Add:", key, ":", value)
```

```python
tea_order("Austine", "chocolate")
```

```python
tea_order("Bob", "black", "oat", "bread", sweetener="honey")
```

```python
tea_order("Tony", "black", milk="oat", sweetener="sugar")
```

```python
# unpacking dictionary into keyword args
eves_extra = {"milk": "almond", "sweetener": "sugar", "flavor": "lemon"}
```

```python
tea_order("Eve", "green", **eves_extra)
```

```python
# unpacking dictionary into positional args
tonys_extra = ("almond", "sugar", "lemon")
```

```python
tea_order("Tony", "black", *tonys_extra)
```
